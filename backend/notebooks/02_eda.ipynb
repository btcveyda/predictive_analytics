# Exploratory Data Analysis

Analyze the California Housing dataset and generate visual insights for the Prophecy AI project.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../app/data/housing.csv')
df.head()
```

```python
df['MedHouseVal'].hist(bins=30)
plt.title('Distribution of Median House Value')
plt.xlabel('Median House Value')
plt.ylabel('Frequency')
plt.show()
```

```python
corr = df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.scatterplot(x='MedInc', y='MedHouseVal', data=df)
plt.title('Median Income vs. Median House Value')
plt.show()
```

```python
sns.scatterplot(x='HouseAge', y='MedHouseVal', data=df)
plt.title('House Age vs. Median House Value')
plt.show()
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(df['Longitude'], df['Latitude'], c=df['MedHouseVal'], cmap='viridis', s=10)
plt.colorbar(label='Median House Value')
plt.title('Geographic Distribution of House Values')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

## Insights

- Median income is typically the strongest predictor of house value.
- House age shows a weaker relationship with house value.
- Geographic clusters appear, with higher values concentrated in specific regions.
